```python
num1 = float(input("Ingresa el primer número: "))
num2 = float(input("Ingresa el segundo número: "))

if num1 > num2:
    print(f"El número mayor es: {num1}")
elif num2 > num1:
    print(f"El número mayor es: {num2}")
else:
    print("Ambos números son iguales.")
```

```python
suma = 0

for i in range(1, 6):
    numero = float(input(f"Ingresa el número {i}: "))
    suma += numero

print(f"\nLa suma total de los 5 números es: {suma}")
```

```python
edad1 = float(input("Ingresa la primera edad: "))
edad2 = float(input("Ingresa la segunda edad: "))
edad3 = float(input("Ingresa la tercera edad: "))

promedio = (edad1 + edad2 + edad3) / 3
print(f"El promedio de las tres edades es: {promedio:.1f}")
```

## Ejercicios
Pida el nombre del estudiante.

Solicite tres calificaciones.

Calcule el promedio.
Muestre el promedio.

Indique si aprobó (70 o más) o reprobó.

```python
# 1. Pedir el nombre del estudiante
nombre = input("Ingresa el nombre del estudiante: ")

# 2. Solicitar las tres calificaciones
calificacion1 = float(input("Ingresa la primera calificación: "))
calificacion2 = float(input("Ingresa la segunda calificación: "))
calificacion3 = float(input("Ingresa la tercera calificación: "))

# 3. Calcular el promedio
promedio = (calificacion1 + calificacion2 + calificacion3) / 3

# 4. Mostrar el promedio
print(f"\nEl promedio de {nombre} es: {promedio:.2f}")

# 5. Indicar si aprobó o reprobó
if promedio >= 70:
    print("Estado: Aprobado")
else:
    print("Estado: Reprobado")
```
